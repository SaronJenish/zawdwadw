## Experiment 3 – Probability Distributions and Bayes’ Theorem

```python
# Generate standard distributions and compare empirical histograms with theoretical PMF/PDF
fig,axs=plt.subplots(2,3,figsize=(13,7))
samples=[
    ("Bernoulli",stats.bernoulli.rvs(.3,size=1000),lambda x:stats.bernoulli.pmf(x,.3),[0,1]),
    ("Binomial",stats.binom.rvs(10,.4,size=1000),lambda x:stats.binom.pmf(x,10,.4),range(11)),
    ("Poisson",stats.poisson.rvs(4,size=1000),lambda x:stats.poisson.pmf(x,4),range(15)),
    ("Uniform",stats.uniform.rvs(0,1,size=1000),lambda x:stats.uniform.pdf(x,0,1),None),
    ("Gaussian",stats.norm.rvs(0,1,size=1000),lambda x:stats.norm.pdf(x),None)]
for ax,(name,s,fn,discrete) in zip(axs.flat,samples):
    if discrete:
        bins=np.arange(min(discrete)-.5,max(discrete)+1.5,1); ax.hist(s,bins=bins,density=True,alpha=.6)
        xx=np.array(list(discrete)); ax.plot(xx,fn(xx),"o-")
    else:
        ax.hist(s,bins=30,density=True,alpha=.6); xx=np.linspace(min(s),max(s),200); ax.plot(xx,fn(xx))
    ax.set_title(name)
axs.flat[-1].axis("off"); plt.tight_layout(); plt.show()
prevalence,sensitivity,specificity=.01,.95,.90
posterior=prevalence*sensitivity/(prevalence*sensitivity+(1-prevalence)*(1-specificity))
print("P(Disease | Positive) =",round(posterior,4))
N=100_000
disease=np.random.rand(N)<prevalence
positive=np.where(disease,np.random.rand(N)<sensitivity,np.random.rand(N)<(1-specificity))
print("Simulated posterior =",positive[disease | ~disease].mean() if False else disease[positive].mean())
p=np.linspace(.001,.20,100)
post=p*sensitivity/(p*sensitivity+(1-p)*(1-specificity))
plt.plot(p*100,post*100); plt.xlabel("Disease prevalence (%)"); plt.ylabel("P(Disease | Positive) (%)"); plt.title("Bayesian posterior vs prevalence"); plt.grid(); plt.show()
```
